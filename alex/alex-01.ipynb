```python
import pandas as pd
import numpy as np
```

```python
g = np.random.default_rng(0)
```

```python
scores = g.integers(70, 101, size=10)
print(scores)
```

```python
s = pd.Series(scores, index=['September', 'October', 'November', 'December', 'January', 'February', 'March', 'April', 'May', 'June'])
print(s)
```

```python
s.mean()
```

```python
s[:5].mean()
```

```python
s[5:].mean()
```

```python
s.iloc[:5].mean()
```

```python
s.loc['September':'January'].mean()
```
